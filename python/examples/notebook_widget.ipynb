# Interactive Spectra plots in Jupyter

`EmbedSurface` renders on the GPU inside the kernel. In a notebook it shows as a live
canvas: **scroll** to zoom, **drag** to pan, **right-drag** to zoom one axis,
**double-click** or press **R** to reset. Works in JupyterLab, Notebook 7, VS Code and Colab.

Needs the notebook extra and a Vulkan driver:

```
pip install "spectra-plot[notebook]"
```

````python
import numpy as np
from spectra._embed import EmbedSurface

x = np.linspace(0, 20, 2000, dtype=np.float32)

surface = EmbedSurface(900, 420)
fig = surface.figure()
ax = fig.subplot(1, 1, 1)
ax.line(x, np.sin(x) * np.exp(-x / 15), label="damped sine")
ax.line(x, np.cos(2 * x) * 0.5, label="cos(2x) / 2")
ax.set_title("Scroll to zoom, drag to pan")
ax.set_xlabel("t [s]")
ax.set_grid(True)
surface  # the last expression shows as a live canvas
````

`spectra.notebook.show()` takes a surface or a figure and returns the widget, so you can set
its height and refresh it after changing the plot from Python. Without anywidget installed,
surfaces fall back to a static PNG. Saved outputs also carry a PNG, so the plot still shows on
GitHub and in nbconvert exports.

````python
import spectra.notebook as snb

surface2 = EmbedSurface(900, 360)
ax2 = surface2.figure().subplot(1, 1, 1)
rng = np.random.default_rng(0)
walk = ax2.line(x, np.cumsum(rng.normal(size=x.size)).astype(np.float32), label="random walk")
canvas = snb.show(surface2, height=300)
````

````python
# Change the data from Python, then push a new frame to the canvas above.
walk.set_y(np.cumsum(rng.normal(size=x.size)).astype(np.float32))
ax2.auto_fit()
canvas.refresh()
````
